```python
import matplotlib.pyplot as plt

n_elements = [1, 5, 10]
displacements = [-152.06, -188.95, -190.11]
exact_displacement = -190.48

plt.plot(n_elements, displacements, marker="o")
plt.axhline(
    y=exact_displacement,
    color="r",
    linestyle="--",
    label=f"Exact Displacement: {exact_displacement:.2f} mm",
)
plt.title("Free end Displacement vs Number of Elements for cantilever beam")
plt.xlabel("Number of Elements")
plt.ylabel("Free end Displacement (mm)")
# annotate the points with their values
offsets = [(20, 3), (20, 5), (-3, 10)]  # offsets for the annotations
for i, txt in enumerate(displacements):
    plt.annotate(
        f"{txt:.2f}",
        (n_elements[i], displacements[i]),
        textcoords="offset points",
        xytext=offsets[i],
        ha="center",
    )
plt.annotate(
    r"$\delta^{\text{theory}}$ ="+f"{exact_displacement:.2f} mm",
    (n_elements[-1], exact_displacement),
    textcoords="offset points",
    xytext=(-280, 10),
    ha="center",
    color="r",
)
plt.grid()
plt.savefig("3_2_comp.png")
plt.show()
```
